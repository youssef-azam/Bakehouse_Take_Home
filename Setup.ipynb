## Schema Creation

```python
CATALOG = "workspace"
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.bakehouse_bronze")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.bakehouse_silver")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.bakehouse_gold")

print(f" Schemas created successfully in catalog '{CATALOG}':")
print(f" - {CATALOG}.bakehouse_bronze")
print(f" - {CATALOG}.bakehouse_silver")
print(f" - {CATALOG}.bakehouse_gold")
```

##Data Exploration

```python
tables = ["sales_transactions", "sales_customers", "sales_franchises"]

for t in tables:
    df = spark.table(f"samples.bakehouse.{t}")
    print(f"\n==========================================")
    print(f" Table: samples.bakehouse.{t} | Rows: {df.count():,}")
    print(f"==========================================")
    df.printSchema()
    display(df.limit(5))
```
